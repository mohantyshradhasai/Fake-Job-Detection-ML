```python
def calculate_ats_score(resume_text, job_description):
    def clean(text):
        text = text.lower()
        text = re.sub(r'[^a-z0-9\s]', '', text)
        return text

    resume_clean = clean(resume_text)
    jd_clean = clean(job_description)

    # ── TF-IDF Cosine Similarity ──────────────────────────────
    vectorizer = TfidfVectorizer()
    vectors = vectorizer.fit_transform([resume_clean, jd_clean])
    cosine_score = cosine_similarity(vectors[0], vectors[1])[0][0] * 100

    # ── Keyword Match ─────────────────────────────────────────
    # Filter stopwords manually for better keyword quality
    stopwords = {"and","the","to","of","in","a","is","for","with","on","are","be","this","that","as","by"}
    jd_keywords  = set(jd_clean.split()) - stopwords
    resume_words = set(resume_clean.split()) - stopwords
    matched      = jd_keywords & resume_words
    missing      = jd_keywords - resume_words

    keyword_score = (len(matched) / len(jd_keywords)) * 100 if jd_keywords else 0

    # ── Section Check ─────────────────────────────────────────
    required_sections = ["experience", "education", "skills", "projects", "summary"]
    found_sections    = [s for s in required_sections if s in resume_clean]
    missing_sections  = [s for s in required_sections if s not in resume_clean]
    section_score     = (len(found_sections) / len(required_sections)) * 100

    # ── Formatting Penalty ────────────────────────────────────
    formatting_penalty = 0
    if len(full_text.strip()) < 200:
        formatting_penalty += 20   # likely unreadable/image-based PDF
    if full_text.count("\n\n\n") > 10:
        formatting_penalty += 10   # too many blank lines = table artifacts

    # ── Weighted Final Score ──────────────────────────────────
    final_score = (
        cosine_score   * 0.40 +
        keyword_score  * 0.35 +
        section_score  * 0.25
    ) - formatting_penalty

    final_score = max(0, min(100, final_score))  # clamp to 0–100

    return {
        "cosine_score"      : round(cosine_score, 2),
        "keyword_score"     : round(keyword_score, 2),
        "section_score"     : round(section_score, 2),
        "formatting_penalty": formatting_penalty,
        "final_ats_score"   : round(final_score, 2),
        "matched_keywords"  : sorted(matched),
        "missing_keywords"  : sorted(missing),
        "found_sections"    : found_sections,
        "missing_sections"  : missing_sections,
    }
```

```python
def generate_gap_advice(ats_result):
    advice = []

    score = ats_result["final_ats_score"]

    # ── Score Band ────────────────────────────────────────────
    if score >= 85:
        band = "🟢 Excellent"
    elif score >= 70:
        band = "🟡 Good — minor tweaks needed"
    elif score >= 50:
        band = "🟠 Average — significant gaps"
    else:
        band = "🔴 Poor — major rework needed"

    print("\n" + "="*55)
    print(f"  ATS SCORE REPORT")
    print("="*55)
    print(f"  Final ATS Score     : {score}%  {band}")
    print(f"  Cosine Similarity   : {ats_result['cosine_score']}%")
    print(f"  Keyword Match       : {ats_result['keyword_score']}%")
    print(f"  Section Coverage    : {ats_result['section_score']}%")
    print(f"  Formatting Penalty  : -{ats_result['formatting_penalty']}%")
    print("="*55)

    # ── Missing Keywords ─────────────────────────────────────
    if ats_result["missing_keywords"]:
        top_missing = ats_result["missing_keywords"][:15]
        print(f"\n❌ Top Missing Keywords ({len(ats_result['missing_keywords'])} total):")
        print("  ", ", ".join(top_missing))
        advice.append(
            f"Add these keywords naturally in your Skills or Experience section:\n"
            f"  👉 {', '.join(top_missing)}"
        )

    # ── Missing Sections ─────────────────────────────────────
    if ats_result["missing_sections"]:
        print(f"\n⚠️  Missing Resume Sections:")
        for s in ats_result["missing_sections"]:
            print(f"   - {s.title()}")
        advice.append(
            f"Add these missing sections with standard ATS headers:\n"
            + "\n".join([f"  👉 Add a '{s.title()}' section" for s in ats_result["missing_sections"]])
        )

    # ── Score-Based General Advice ────────────────────────────
    if score < 50:
        advice.append("📌 Tailor your resume specifically to this job description — don't use a generic resume.")
        advice.append("📌 Use the exact phrases from the job posting, not paraphrased versions.")
    if score < 70:
        advice.append("📌 Add a dedicated 'Skills' section listing all technical and soft skills.")
        advice.append("📌 Quantify achievements: use numbers, %, $, or time saved.")
    if ats_result["formatting_penalty"] > 0:
        advice.append("📌 Fix formatting: avoid tables, columns, or image-based text — use plain single-column layout.")

    print("\n💡 Improvement Advice:")
    for i, tip in enumerate(advice, 1):
        print(f"\n  {i}. {tip}")

    print("\n✅ Matched Keywords Sample:")
    print("  ", ", ".join(sorted(ats_result["matched_keywords"])[:15]))
    print("="*55)
```
